# Pandas for Data Science
Easy notes and runnable examples.

## Import Pandas

```python
import pandas as pd
import matplotlib.pyplot as plt 
df=pd.DataFrame({'Name':['Alice','Bob','Charlie','David'],'Age':[20,22,21,23],'City':['Dhaka','Khulna','Dhaka','Rajshahi'],'Marks':[90,75,85,None]})
df
```

## Inspection

```python
print(df.shape)
print(df.columns)
df.info()
display(df.head())
display(df.tail())
display(df.sample(2))
display(df.describe(include='all'))
```

## Rename

```python
df.rename(columns={'Marks':'Score'})
```

## Set Index

```python
df.set_index('Name')
```

## Select Columns

```python
display(df['Age'])
display(df[['Name','City']])
```

## iloc and loc

```python
display(df.iloc[0:2])
display(df.loc[0:2,['Name','Marks']])
```

## Filtering

```python
display(df[df['Age']>20])
display(df[df['City'].isin(['Dhaka','Khulna'])])
display(df[df['Age'].between(20,22)])
```

## String Filters

```python
display(df[df['Name'].str.startswith('A')])
display(df[df['Name'].str.contains('a',case=False)])
```

## Missing Values

```python
display(df[df['Marks'].isna()])
display(df[df['Marks'].notna()])
```

## Sorting

```python
df.sort_values('Age',ascending=False)
```

## GroupBy

```python
df.groupby('City')['Age'].mean()
```
