# Basic Python for Data Science
Easy notes and runnable examples.

## Variables

```python
a=10
name='Alice'
print(a,name)
```

## Data Types

```python
x=10
y=3.5
s='Hi'
b=True
print(type(x),type(y),type(s),type(b))
```

## Operators

```python
print("Adddition:",5+2)
print("Subtraction:",5-2)
print("Multiplication:",5*2)
print("Division:",5/2)
print("Floor Division:",5//2)
print("Modulus:",5%2)
print("Exponentiation:",2**3)
```

## If Else

```python
marks=75
if marks>=80:
 print('A')
elif marks>=70:
 print('B')
else:
 print('C')
```

## For Loop

```python
for i in range(5):
 print(i)
```

## While Loop

```python
i=1
while i<=3:
 print(i); i+=1
```

## Lists

```python
nums=[1,2,3]
nums.append(4)
print(nums, nums[0], nums[1:3])
```

## Tuple

```python
t=(1,2,3)
print(t)
```

## Dictionary

```python
d={'Name':'Alice','Age':20}
print(d['Name'])
```

## Functions

```python
def add(a,b): return a+b
print(add(2,3))
```

## Strings

```python
s='Alice'
print(s.upper(),s.lower(),s.replace('A','E'))
```
